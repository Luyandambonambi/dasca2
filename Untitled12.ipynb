```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the dataset with correct parameters
df = pd.read_csv(r"C:\Users\luyanda\Desktop\New folder\School_Data_CSV.csv", header=None, encoding='latin-1', low_memory=False)

# Remove the first row which contains the original column names
df = df.iloc[1:].reset_index(drop=True)

# Rename only the columns we need using the correct indices
df = df.rename(columns={4: 'quintile', 5: 'ward_id', 7: 'new_long', 8: 'new_lat'})

# Coerce the numeric columns
df['quintile'] = pd.to_numeric(df['quintile'], errors='coerce')
df['ward_id'] = pd.to_numeric(df['ward_id'], errors='coerce')
df['new_long'] = pd.to_numeric(df['new_long'], errors='coerce')
df['new_lat'] = pd.to_numeric(df['new_lat'], errors='coerce')

# Drop missing values
df = df.dropna(subset=['quintile', 'ward_id'])

# Take a random sample of 10000 rows to avoid memory errors
df = df.sample(n=10000, random_state=42)

# Group by ward and quintile
ward_quintile_counts = df.groupby(['ward_id', 'quintile']).size().unstack(fill_value=0)

# Plot
ward_quintile_counts.plot(kind='line', figsize=(12, 6))
plt.title('Number of Schools per Quintile Across Wards')
plt.xlabel('Ward')
plt.ylabel('Number of Schools')
plt.legend(title='Quintile')
plt.grid(True)
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the dataset and assign column names directly
df = pd.read_csv(r"C:\Users\luyanda\Desktop\New folder\School_Data_CSV.csv", header=None, encoding='latin-1', low_memory=False)
df.columns = ['emisno', 'centrename', 'institution_name', 'centro', 'quintile', 'ward_id', 'street_address', 'new_long', 'new_lat']

# Drop the first row that contained the original headers
df = df.iloc[1:].reset_index(drop=True)

# Convert numeric columns
df['quintile'] = pd.to_numeric(df['quintile'], errors='coerce')
df['ward_id'] = pd.to_numeric(df['ward_id'], errors='coerce')

# Drop rows with missing critical data and take a sample to avoid memory issues
df = df.dropna(subset=['quintile', 'ward_id']).sample(n=10000, random_state=42)

# Set the visual theme
sns.set_theme(style="whitegrid")

# Create the line plot
plt.figure(figsize=(12, 6))
sns.lineplot(data=df, x='ward_id', y=df.groupby(['ward_id', 'quintile'])['emisno'].transform('count'), hue='quintile', errorbar=None)
plt.title('Number of Schools per Quintile Across Wards')
plt.xlabel('Ward')
plt.ylabel('Number of Schools')
plt.legend(title='Quintile')
plt.show()
```

```python
import pandas as pd
import plotly.express as px

# Load and clean the data
df = pd.read_csv(r"C:\Users\luyanda\Desktop\New folder\School_Data_CSV.csv", header=None, encoding='latin-1', low_memory=False)
df.columns = ['emisno', 'centrename', 'institution_name', 'centro', 'quintile', 'ward_id', 'street_address', 'new_long', 'new_lat']
df = df.iloc[1:].reset_index(drop=True)

# Convert coordinates to numeric
df['new_long'] = pd.to_numeric(df['new_long'], errors='coerce')
df['new_lat'] = pd.to_numeric(df['new_lat'], errors='coerce')
df['quintile'] = pd.to_numeric(df['quintile'], errors='coerce')

# Drop missing and sample
df = df.dropna(subset=['new_long', 'new_lat']).sample(n=5000, random_state=42)

# Create the interactive scatter plot
fig = px.scatter(df, x='new_long', y='new_lat', color='quintile',
                 title='Geographic Locations of Schools',
                 labels={'new_long': 'Longitude', 'new_lat': 'Latitude'})
fig.show()
```
